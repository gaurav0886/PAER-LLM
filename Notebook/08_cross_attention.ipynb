# Cross-Attention Fusion

## Objective

This notebook implements the attention mechanism used to fuse psychoacoustic and HuBERT representations.

The implementation follows the Scaled Dot Product Attention introduced in the Transformer architecture and will later be extended to Cross-Attention for multimodal emotion recognition.


```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
from Fusion.cross_attention import CrossAttention
```

```python
torch.manual_seed(42)

query = torch.rand(4, 8)

key = torch.rand(4, 8)

value = torch.rand(4, 8)

print(query.shape)
print(key.shape)
print(value.shape)
```

```python
attention_layer = CrossAttention(
    input_dim=8,
    hidden_dim=16
)
```

```python
output, attention = attention_layer(
    query,
    key,
    value
)

print(output.shape)
print(attention.shape)
```
