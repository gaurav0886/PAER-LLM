# Unified Psychoacoustic Representation

## Objective

This notebook constructs the unified psychoacoustic feature representation proposed in this research.

The representation combines multiple psychoacoustic descriptors including:

- MFCC
- Pitch
- Loudness
- Spectral
- Bark

into a single feature vector for each speech utterance.


```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)

import pandas as pd
import numpy as np
```

```python
pitch = pd.read_csv("../Outputs/pitch_features.csv")
loudness = pd.read_csv("../Outputs/loudness_features.csv")
spectral = pd.read_csv("../Outputs/spectral_features.csv")
bark = pd.read_csv("../Outputs/bark_features.csv")
```

```python
print("Pitch :", pitch.shape)

print("Loudness :", loudness.shape)

print("Spectral :", spectral.shape)

print("Bark :", bark.shape)
```

```python
pitch["file"].equals(loudness["file"])
```

```python
pitch["file"].equals(spectral["file"])
```

```python
pitch["file"].equals(bark["file"])
```

```python
fusion = pd.DataFrame()

fusion["file"] = pitch["file"]

fusion["emotion"] = pitch["emotion"]
```

```python
pitch_columns = [
    c for c in pitch.columns
    if c not in ["file", "emotion"]
]

fusion = pd.concat(
    [
        fusion,
        pitch[pitch_columns]
    ],
    axis=1
)
```

```python
loudness_columns = [
    c for c in loudness.columns
    if c not in ["file", "emotion"]
]

fusion = pd.concat(
    [
        fusion,
        loudness[loudness_columns]
    ],
    axis=1
)
```

```python
bark_columns = [
    c for c in bark.columns
    if c not in ["file", "emotion"]
]

fusion = pd.concat(
    [
        fusion,
        bark[bark_columns]
    ],
    axis=1
)
```
