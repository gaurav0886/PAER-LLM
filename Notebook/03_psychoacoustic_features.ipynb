```python
import os
import sys

# Move one level up from Notebook to the project root
project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
import librosa
import numpy as np
import pandas as pd

from tqdm import tqdm

import parselmouth
from Features.pitch import extract_pitch_features
from Features.loudness import extract_loudness_features
from Features.spectral import extract_spectral_features
```

```python
emotion_map = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}
```

```python
dataset_path = r"F:\PhD\PhD_Project\Dataset"

sample_file = None

for root, dirs, files in os.walk(dataset_path):

    for file in files:

        if file.endswith(".wav"):

            sample_file = os.path.join(
                root,
                file
            )

            break

    if sample_file:
        break

sample_file
```

```python
features = extract_pitch_features(
    sample_file
)

features
```

```python
pitch_data = []

for root, dirs, files in os.walk(dataset_path):

    for file in tqdm(files):

        if file.endswith(".wav"):

            file_path = os.path.join(
                root,
                file
            )

            emotion_id = file.split("-")[2]

            emotion = emotion_map[
                emotion_id
            ]

            features = extract_pitch_features(
                file_path
            )

            pitch_data.append(

                [
                    file
                ]

                +

                features

                +

                [
                    emotion
                ]

            )
```

```python
columns = [

    "file",

    "mean_pitch",

    "max_pitch",

    "min_pitch",

    "std_pitch",

    "pitch_range",

    "emotion"

]

pitch_df = pd.DataFrame(
    pitch_data,
    columns=columns
)

pitch_df.head()
```

```python
output_path = r"F:\PhD\PhD_Project\Outputs\pitch_features.csv"

pitch_df.to_csv(
    output_path,
    index=False
)

print(
    "Saved Successfully"
)
```

# Spectral Feature Extraction

```python
extract_spectral_features(sample_file)
```

```python
spectral_data = []

for root, dirs, files in os.walk(dataset_path):

    for file in tqdm(files):

        if file.endswith(".wav"):

            file_path = os.path.join(root, file)

            emotion_id = file.split("-")[2]
            emotion = emotion_map[emotion_id]

            features = extract_spectral_features(file_path)

            spectral_data.append(
                [file] + features + [emotion]
            )
```

```python
columns = [

    "file",

    "spectral_centroid",

    "spectral_bandwidth",

    "spectral_rolloff",

    "spectral_contrast",

    "spectral_flatness",

    "emotion"

]

spectral_df = pd.DataFrame(
    spectral_data,
    columns=columns
)

spectral_df.head()
```

```python
output_path = r"F:\PhD\PhD_Project\Outputs\spectral_features.csv"

spectral_df.to_csv(
    output_path,
    index=False
)

print("Spectral Features Saved Successfully!")
```
