```python
import os
import sys

# Move one level up from Notebook to the project root
project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert

dataset_path = r"F:\PhD\PhD_Project\Dataset"

total_files = 0

for root, dirs, files in os.walk(dataset_path):
    for file in files:
        if file.endswith(".wav"):
            total_files += 1

print("Total Audio Files:", total_files)
```

```python
emotion_map = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}

print(emotion_map)
```

```python
import os
from collections import Counter

emotion_counts = Counter()

for root, dirs, files in os.walk(dataset_path):

    for file in files:

        if file.endswith(".wav"):

            emotion_id = file.split("-")[2]

            emotion_name = emotion_map[emotion_id]

            emotion_counts[emotion_name] += 1

print(emotion_counts)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(
    emotion_counts.keys(),
    emotion_counts.values()
)

plt.title("Emotion Distribution")
plt.xlabel("Emotion")
plt.ylabel("Count")

plt.show()
```

```python
import librosa

sample_file = None

for root, dirs, files in os.walk(dataset_path):

    for file in files:

        if file.endswith(".wav"):
            sample_file = os.path.join(root,file)
            break

    if sample_file:
        break

audio, sr = librosa.load(sample_file)

print("Sampling Rate:", sr)
print("Duration:", len(audio)/sr)
```

```python
import librosa.display

plt.figure(figsize=(12,4))

librosa.display.waveshow(audio,sr=sr)

plt.title("Speech Waveform")

plt.show()
```
