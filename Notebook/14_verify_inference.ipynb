```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)

from Utils.inference import (
    load_emotion_model,
    predict
)
```

```python
emotion_model = load_emotion_model()
```

```python
emotion_map = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}
```

```python
results = []

dataset_path = "../Dataset"

for actor in sorted(os.listdir(dataset_path)):

    actor_path = os.path.join(dataset_path, actor)

    if not os.path.isdir(actor_path):
        continue

    for file in sorted(os.listdir(actor_path)):

        if not file.endswith(".wav"):
            continue

        audio_path = os.path.join(actor_path, file)

        emotion_code = file.split("-")[2]

        true_emotion = emotion_map[emotion_code]

        predicted, confidence, embedding = predict(
            audio_path,
            emotion_model
        )

        results.append({

            "file": file,

            "true_emotion": true_emotion,

            "predicted_emotion": predicted,

            "confidence": confidence

        })

        print(file, "->", predicted)
```

```python
import pandas as pd
results_df = pd.DataFrame(results)

results_df.head()
```

```python
results_df.to_csv(
    "../Outputs/inference_results.csv",
    index=False
)

print("Saved")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

print(
    "Accuracy:",
    accuracy_score(
        results_df.true_emotion,
        results_df.predicted_emotion
    )
)

print()

print(

    classification_report(

        results_df.true_emotion,

        results_df.predicted_emotion

    )

)
```
