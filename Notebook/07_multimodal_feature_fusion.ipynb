# Multimodal Feature Fusion

## Objective

This notebook constructs the multimodal representation by integrating:

- Psychoacoustic Features
- HuBERT Embeddings

The fused representation will be used for emotion recognition.

```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)

import pandas as pd
import numpy as np
import joblib

from sklearn.preprocessing import StandardScaler

```

```python
psycho = pd.read_csv(
    "../Outputs/psychoacoustic_features.csv"
)

hubert = pd.read_csv(
    "../Outputs/hubert_embeddings.csv"
)
```

```python
print(psycho.shape)

print(hubert.shape)
```

```python
psycho["file"].equals(
    hubert["file"]
)
```

```python
meta = psycho[
    ["file","emotion"]
]

psycho_features = psycho.drop(
    columns=["file","emotion"]
)

hubert_features = hubert.drop(
    columns=["file","emotion"]
)
```

```python
psycho_scaler = StandardScaler()
hubert_scaler = StandardScaler()

psycho_scaled = psycho_scaler.fit_transform(
    psycho_features
)

hubert_scaled = hubert_scaler.fit_transform(
    hubert_features
)

# Save the scalers
joblib.dump(
    psycho_scaler,
    "../Models/psycho_scaler.pkl"
)

joblib.dump(
    hubert_scaler,
    "../Models/hubert_scaler.pkl"
)

print("Scalers Saved Successfully")
```

```python
fusion = np.concatenate(
    [
        psycho_scaled,
        hubert_scaled
    ],
    axis=1
)
```

```python
columns = (
    list(psycho_features.columns)
    +
    list(hubert_features.columns)
)

fusion_df = pd.DataFrame(
    fusion,
    columns=columns
)

fusion_df.insert(
    0,
    "emotion",
    meta["emotion"]
)

fusion_df.insert(
    0,
    "file",
    meta["file"]
)
```

```python
print(fusion_df.shape)

fusion_df.head()
```

```python
fusion_df.to_csv(
    "../Outputs/multimodal_features.csv",
    index=False
)

print("Multimodal Dataset Saved")
```
